```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
data_path = "C:/Assignment 6/Supermart Grocery Sales - Retail Analytics Dataset.csv"
sales_data = pd.read_csv(data_path)
```

```python
# Check the column names to ensure there are no mismatches
print("Column names: ", sales_data.columns)
```

```python
# Strip any extra spaces from column names, if any
sales_data.columns = sales_data.columns.str.strip()
```

```python
# Convert 'Order Date' to datetime (handle mixed date formats)
sales_data['Order Date'] = pd.to_datetime(sales_data['Order Date'], errors='coerce')
```

```python
# Check for any invalid date entries
print("Invalid Date Entries (NaT): ", sales_data[sales_data['Order Date'].isna()])
```

```python
# Remove rows with missing values (including dates)
sales_data = sales_data.dropna()
```

```python
# Create a Profit-to-Sales column
sales_data['Profit_To_Sales'] = sales_data['Profit'] / sales_data['Sales']
```

```python
# Visualizations: Total Sales by Category (Bar Chart)
category_sales = sales_data.groupby('Category').Sales.sum().reset_index()
```

```python
# Bar Chart for Total Sales by Category
plt.figure(figsize=(10, 6))
sns.barplot(x='Category', y='Sales', data=category_sales, palette='muted')
plt.title('Total Sales by Category')
plt.xticks(rotation=45)
plt.ylabel('Total Sales')
plt.xlabel('Category')
plt.show()
```

```python
# Sales vs Profit (Scatter Plot)
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Sales', y='Profit', hue='Category', data=sales_data, palette='muted')
plt.title('Sales vs Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.show()
```
